```python
# Takes as input audio files in the folder "demo", split them into random 3s tracks and
# puts them in the folder "demo_3s"

import os
import random
from pydub import AudioSegment

input_dir = "../audio/demo"
output_dir = "../audio/demo_3s"
os.makedirs(output_dir, exist_ok=True)

num_samples = 20      
sample_duration = 3*1000  

for filename in os.listdir(input_dir):
    if not filename.lower().endswith((".wav", ".mp3", ".flac", ".ogg")):
        continue
    
    filepath = os.path.join(input_dir, filename)
    
    audio = AudioSegment.from_file(filepath)
    total_duration = len(audio)  
    
    if total_duration <= sample_duration:
        print(f"{filename} too short ({total_duration/1000:.2f}s), skipped.")
        continue
    
    for i in range(1, num_samples + 1):
        max_start = total_duration - sample_duration
        start = random.randint(0, max_start)
        end = start + sample_duration
        segment = audio[start:end]
        
        base_name, _ = os.path.splitext(filename)
        out_name = f"{base_name}_{i}.wav"
        out_path = os.path.join(output_dir, out_name)
        segment.export(out_path, format="wav")
    
    print(f"Processed {filename}, created {num_samples} segments.")

print("Completed!")
```
